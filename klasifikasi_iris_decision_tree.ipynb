## 2. Import library

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay
)

print("Library berhasil diimpor.")
```

## 3. Memuat dataset Iris langsung dari scikit-learn
Data dimuat langsung ke memori. Tidak ada file CSV yang dibaca.

```python
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target
nama_kelas = list(iris.target_names)

df = X.copy()
df["species"] = y.map(dict(enumerate(nama_kelas)))

print("Jumlah baris:", df.shape[0])
print("Jumlah fitur:", X.shape[1])
print("\nLima data pertama:")
display(df.head())
print("\nJumlah data per kelas:")
display(df["species"].value_counts())
```

## 4. Eksplorasi data
Grafik sebaran panjang dan lebar petal untuk tiga kelas bunga.

```python
plt.figure(figsize=(8, 5))
for kelas in nama_kelas:
    bagian = df[df["species"] == kelas]
    plt.scatter(
        bagian["petal length (cm)"],
        bagian["petal width (cm)"],
        label=kelas,
        alpha=0.8
    )
plt.xlabel("Panjang petal (cm)")
plt.ylabel("Lebar petal (cm)")
plt.title("Sebaran Petal pada Dataset Iris")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## 5. Membagi data training dan testing
80% untuk training dan 20% untuk testing. `stratify=y` menjaga proporsi kelas.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Data training:", len(X_train))
print("Data testing :", len(X_test))
```

## 6. Membuat dan melatih model Decision Tree

```python
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
print("Model Decision Tree berhasil dilatih.")
```

## 7. Evaluasi model
Akurasi, precision, recall, F1-score, dan confusion matrix.

```python
y_pred = model.predict(X_test)
akurasi = accuracy_score(y_test, y_pred)

print(f"Akurasi: {akurasi:.2%}")
print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=nama_kelas, digits=3))

cm = confusion_matrix(y_test, y_pred)
print("Confusion matrix:")
print(cm)
```

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=nama_kelas)
disp.plot(values_format="d")
plt.title("Confusion Matrix - Iris Decision Tree")
plt.tight_layout()
plt.show()
```

## 8. Visualisasi pohon keputusan

```python
plt.figure(figsize=(15, 8))
plot_tree(
    model,
    feature_names=list(X.columns),
    class_names=nama_kelas,
    filled=True,
    rounded=True,
    fontsize=9
)
plt.title("Pohon Keputusan untuk Klasifikasi Bunga Iris")
plt.tight_layout()
plt.show()
```

## 9. Prediksi bunga baru
Urutan fitur: sepal length, sepal width, petal length, petal width (cm).

```python
bunga_baru = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=X.columns
)
prediksi = model.predict(bunga_baru)[0]
probabilitas = model.predict_proba(bunga_baru)[0]

display(bunga_baru)
print("Prediksi jenis bunga:", nama_kelas[prediksi])
print("\nProbabilitas model:")
for nama, nilai in zip(nama_kelas, probabilitas):
    print(f"{nama}: {nilai:.1%}")
```

## 10. Kesimpulan
Model Decision Tree dilatih menggunakan empat fitur numerik bunga Iris. Evaluasi dilakukan pada data testing yang tidak digunakan saat pelatihan. Gunakan nilai akurasi dan confusion matrix yang muncul saat notebook dijalankan untuk menyusun kesimpulan akhir.
